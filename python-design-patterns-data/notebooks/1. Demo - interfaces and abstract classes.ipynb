
Demostrating Interfaces and Abstract classes 

A "contract" is like a promise to provide some specific behaviour or functionality. In programming, 
we can use interfaces and abstract classes to define these contracts.


This program demonstrates the use of abstract classes and method overriding in Python. 
We define an abstract class 'Animal' with an abstract method 'sound' and a concrete method 'description'. 
The 'Dog' class overrides both the 'sound' and 'description' methods, while the 'Cat' class only overrides 
the 'sound' method, inheriting the default implementation of 'description' from the 'Animal' class. 

When we create instances of 'Dog' and 'Cat', we see how the overridden methods behave differently for each class.



```python
from abc import ABC, abstractmethod
```

# interface contract - any class that inherits from 'Animal' must implement the 'sound' method, and can optionally 
# override the 'description' method.

```python

# Define an abstract class 'Animal'
# we enforce the contract here by typehinting i.e. by specifying @abstractmethod and the method signature
# interface contract - children must implement the method 'sound' and 'description', i.e. must implement all of the 
# abstract methods defined in the parent class
# note that these methods have no implementation in the parent class, they are just defined as abstract methods 
# that must be implemented by any concrete subclass

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

    # Make the 'description' method abstract but provide a basic implementation
    def description(self):
        print(f"{self.__class__.__name__} says: {self.sound()}")

# Define a concrete class 'Dog' that inherits from 'Animal'
class Dog(Animal):
    def sound(self):
        return "Woof!"
    
    def description(self):
        print(f"My little dog says: {self.sound()}")

# Define a concrete class 'Cat' that inherits from 'Animal'
class Cat(Animal):
    def sound(self):
        return "Meow!"

# Create instances of concrete classes and use the overridden 'description' method
dog = Dog()
dog.description()

cat = Cat()
cat.description()
```

Why do we need abstract classes ?

Abstract classes are essential in object-oriented programming for several reasons:

    Cannot be Instantiated: Abstract classes cannot be instantiated directly. They serve as blueprints for other classes. This ensures that specific implementation details are provided by subclass implementations.

    Defining Contracts: An abstract class can define a contract that derived classes must follow. Any class inheriting from an abstract class is required to implement the abstract methods, ensuring that they adhere to a defined interface.

    Promotes Code Reusability: Abstract classes can contain both abstract methods (without implementation) and concrete methods (with implementation). This allows subclasses to inherit common functionality, while also being enforced to provide their specific behavior for the abstract methods. This promotes code reusability and reduces duplication.

    Partial Implementation: They allow for a combination of implemented and abstract methods, providing a base-level implementation while still enforcing structure in the derived classes.

    Illustration of Relationships: Abstract classes help in the organization of related classes within a hierarchy, making it easier to manage complex object relationships.

In summary, abstract classes provide a way to enforce a consistent interface, promote code reuse, and define shared behaviors among a group of related classes. This plays a crucial role in designing scalable and maintainable software systems, particularly when applying design patterns.


```python
from abc import ABC, abstractmethod

class MyInterface(ABC):
    @abstractmethod
    def my_method(self):
        pass

class MyClass (MyInterface):
    def my_method(self):
        print("MyClass implementation of my_method")

class NotImplementingInterface:
    def some_other_method(self):
        print("This class does not implement MyInterface")

# This method expects to be passed an object that implements MyInterface methods
# if we try to pass an object that does not implement MyInterface, it will raise an error
# this is because the method is expecting an object that has the 'my_method' method defined, 
# and if we pass an object that does not have this method, it will raise an AttributeError        

def process_my_interface_object(obj: MyInterface):
    obj.my_method()
    print ("Correct implementation of MyInterface")

my_obj = MyClass()
process_my_interface_object(my_obj)
```

```python
# Abstract classes are in between interfaces and concrete classes. They can have both abstract methods (which must be implemented by subclasses) and concrete methods (which have an implementation in the abstract class).
# Abstract classes can also have instance variables and constructors, which interfaces cannot have.

from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self, color):
        self.color = color

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

    # This method is abstract but has a default implementation. Subclasses can override it, but they are not required to.
    @abstractmethod
    def description(self):
        print (f"This is a {self.color} shape with an area of {self.area()} and a perimeter of {self.perimeter()}")

    # Note: Of the above three abstract methods, only 'area' and 'perimeter' are truly abstract, as they have no implementation 
    # in the abstract class. The 'description' method is abstract in the sense that it is decorated with @abstractmethod, 
    # but it has a default implementation that can be used by subclasses if they choose not to override it.

class Rectangle(Shape):
    def __init__(self, color, width, height):
        # Call the constructor of the abstract class to initialize the 'color' instance variable
        super().__init__(color)
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    # We can choose to override the 'description' method, but we are not required to since it has a default implementation in the abstract class.
    def description(self):
        print (f"This is a {self.color} rectangle with an area of {self.area()} and a perimeter of {self.perimeter()}")

rectangle = Rectangle("red", 4, 5)
rectangle.description()
print (f"Area: {rectangle.area()}")
print (f"Perimeter: {rectangle.perimeter()}")   
```
